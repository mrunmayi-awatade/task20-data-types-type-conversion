```python
# Creating a sample DataFrame
import pandas as pd
data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Neha"],
    "Age": ["21", "22", "20", "23", "21"],
    "Marks": ["85.5", "90.0", "78.5", "92.0", "88.0"],
    "Passed": [1, 1, 0, 1, 1],
    "Date": ["2026-10-01", "2026-10-02", "2026-10-03", "2026-10-04", "2026-10-05"]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Checking data types of each column
print(df.dtypes)
```

```python
# Displaying DataFrame information
df.info()
```

```python
# Converting Name column to string
df["Name"] = df["Name"].astype(str)
print(df.dtypes)
```

```python
# Converting Age column to integer
df["Age"] = df["Age"].astype(int)
print(df[["Age"]])
print(df.dtypes)
```

```python
# Converting Marks column to float
df["Marks"] = df["Marks"].astype(float)
print(df[["Marks"]])
print(df.dtypes)
```

```python
# Converting Passed column to boolean
df["Passed"] = df["Passed"].astype(bool)
print(df[["Passed"]])
print(df.dtypes)
```

```python
# Converting Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])
print(df[["Date"]])
print(df.dtypes)
```

```python
# Converting Age column to numeric values
df["Age"] = pd.to_numeric(df["Age"])
print(df["Age"])
print(df.dtypes)
```

```python
# Converting invalid numeric values into NaN
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")
print(df["Marks"])
```

```python
# Checking final data types
print(df.dtypes)
```

```python
# Displaying the final dataset
print(df)
```
